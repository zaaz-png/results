```python
from pandas.api.types import CategoricalDtype

# Define the custom order for the blood pressure groups
bp_group_order = ['50-63', '63-76', '76-89', '89-102', '102-115', '115-128', '128-141', '141-154', '154-167', '167-180']

# Create a CategoricalDtype with the specified order
cat_type = CategoricalDtype(categories=bp_group_order, ordered=True)

# Apply the categorical type to the 'Grouped BP' column
data_grp['Grouped BP'] = data_grp['Grouped BP'].astype(cat_type)

# Sort the DataFrame by the 'Grouped BP' column
data_grp_asc = data_grp.sort_values(by='Grouped BP')

display(data_grp_asc)
```

```python
# Reset the index of the sorted DataFrame
data_grp_asc = data_grp_asc.reset_index(drop=True)
display(data_grp_asc)
```

```python
data_grp_asc['lower limit']= data_grp_asc['Grouped BP'].str.split('-').str[0].astype(int)
data_grp_asc['upper limit']= data_grp_asc['Grouped BP'].str.split('-').str[1].astype(int)
data_grp_asc
```

```python
# Calculate Inter-Quartile Range (IQR) for grouped data

total_frequency = data_grp_asc['Frequency'].sum()

# Calculate Q1
q1_position = total_frequency / 4
cum_freq = 0
q1_lower_limit = 0
q1_frequency = 0
q1_cf_before = 0

for i in range(len(data_grp_asc)):
    cum_freq += data_grp_asc['Frequency'][i]
    if cum_freq >= q1_position:
        q1_lower_limit = data_grp_asc['lower limit'][i]
        q1_frequency = data_grp_asc['Frequency'][i]
        # The cumulative frequency of the class preceding the quartile class
        if i > 0:
            q1_cf_before = cum_freq - data_grp_asc['Frequency'][i]
        else:
            q1_cf_before = 0
        break

q1 = q1_lower_limit + ((q1_position - q1_cf_before) / q1_frequency) * class_width
print(f'Q1 for grouped data: {q1}')

# Calculate Q3
q3_position = 3 * total_frequency / 4
cum_freq = 0
q3_lower_limit = 0
q3_frequency = 0
q3_cf_before = 0

for i in range(len(data_grp_asc)):
    cum_freq += data_grp_asc['Frequency'][i]
    if cum_freq >= q3_position:
        q3_lower_limit = data_grp_asc['lower limit'][i]
        q3_frequency = data_grp_asc['Frequency'][i]
        # The cumulative frequency of the class preceding the quartile class
        if i > 0:
            q3_cf_before = cum_freq - data_grp_asc['Frequency'][i]
        else:
            q3_cf_before = 0
        break

q3 = q3_lower_limit + ((q3_position - q3_cf_before) / q3_frequency) * class_width
print(f'Q3 for grouped data: {q3}')

# Calculate IQR
iqr_grouped = q3 - q1
print(f'Inter-quartile range for grouped data: {iqr_grouped}')
```
