```python
import pandas as pd
```

```python
import numpy as np
```

```python
from google.colab import files
uploaded=files.upload()
data=pd.read_csv('dataset2.csv')
```

```python
import math
```

```python
import matplotlib.pyplot as plt
```

```python
data['Pus cells clumps'].fillna(data['Pus cells clumps'].mode()[0], inplace=True)
```

```python
data['Bacteria'].fillna(data['Bacteria'].mode()[0], inplace=True)
```

```python
data['Serum Creatinine'].fillna(data['Serum Creatinine'].mean(), inplace=True)
```

```python
data['Apetite'].fillna(data['Apetite'].mode()[0], inplace=True)
```

```python
data['Chronic kidney failure'].fillna(data['Chronic kidney failure'].mode()[0], inplace=True)
```

```python
#finding correlation between Hemoglobin and Random blood glucose
numerator=0
deno1=0
deno2=0
for i in range(0, data.shape[0]):
  numerator+= (data['Hemoglobin'][i]-data['Hemoglobin'].mean())*(data['Random blood glucose'][i]-data['Random blood glucose'].mean())
  deno1 += (data['Hemoglobin'][i]-data['Hemoglobin'].mean())**2
  deno2+= (data['Random blood glucose'][i]-data['Random blood glucose'].mean())**2

denominator= math.sqrt(deno1*deno2)
corr_coeff=numerator/denominator
print(corr_coeff)
```

```python
data.plot.scatter(x='Hemoglobin', y='Random blood glucose', title='Correlation between Hemoglobin and Random blood glucose', alpha=0.1)
plt.show()
```

 AGE and SUGAR IN URINE are slightly negatively correlated


```python
#finding correlation between Packed corpuscular volume and Urine specific gravity
numerator=0
deno1=0
deno2=0
for i in range(0, data.shape[0]):
  numerator+= (data['Packed corpuscular volume'][i]-data['Packed corpuscular volume'].mean())*(data['RBC count'][i]-data['RBC count'].mean())
  deno1 += (data['Packed corpuscular volume'][i]-data['Packed corpuscular volume'].mean())**2
  deno2+= (data['RBC count'][i]-data['RBC count'].mean())**2

denominator= math.sqrt(deno1*deno2)
corr_coeff2=numerator/denominator
print(corr_coeff2)
```

```python
data.plot.scatter(x='Packed corpuscular volume', y='RBC count', title='Correlation between Packed corpuscular volume and RBC count', alpha=0.1)
plt.show()
```

 Albumin in urine and Urine specific gravity are  positively  correlated

```python
#finding correlation between AGE and Sodium in blood
numerator=0
deno1=0
deno2=0
for i in range(0, data.shape[0]):
  numerator+= (data['Age'][i]-data['Age'].mean())*(data['Sodium in blood'][i]-data['Sodium in blood'].mean())
  deno1 += (data['Age'][i]-data['Age'].mean())**2
  deno2+= (data['Sodium in blood'][i]-data['Sodium in blood'].mean())**2

denominator= math.sqrt(deno1*deno2)
corr_coeff=numerator/denominator
print(corr_coeff)
```

```python
data.plot.scatter(x='Age', y='Sodium in blood', title='Correlation between age and Sodium in blood', alpha=0.1)
plt.show()
```

 AGE and Sodium in blood are almost not correlated
